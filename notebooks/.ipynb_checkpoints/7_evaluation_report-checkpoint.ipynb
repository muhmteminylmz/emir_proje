```python
import sys
sys.path.append('..')

import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, roc_auc_score
from src.utils import load_model, print_section, FIGURES_DIR

print('✅ Import başarılı')
```

```python
# Tüm verileri yükle
X_train = np.load('../data/processed/X_train.npy')
X_test  = np.load('../data/processed/X_test.npy')
y_train = np.load('../data/processed/y_train.npy')
y_test  = np.load('../data/processed/y_test.npy')
X_adv_minimal  = np.load('../data/processed/X_adv_minimal.npy')
X_adv_gaussian = np.load('../data/processed/X_adv_gaussian.npy')
X_adv_combined = np.load('../data/processed/X_adv_combined.npy')

with open('../data/processed/feature_names.json') as f:
    feature_names = json.load(f)

# Modelleri yükle
baseline_models = {
    'RF':   load_model('baseline_random_forest'),
    'XGB':  load_model('baseline_xgboost'),
    'LR':   load_model('baseline_logistic_regression'),
}

try:
    defended_models = {
        'RF-D':  load_model('defended_random_forest_robust'),
        'XGB-D': load_model('defended_xgboost_robust'),
        'LR-D':  load_model('defended_logistic_regression_robust'),
        'ENS':   load_model('defended_ensemble'),
    }
except:
    defended_models = {}
    print('⚠️  Savunmalı modeller bulunamadı, 5_defense_mechanism.ipynb çalıştırın')

all_models = {**baseline_models, **defended_models}
print(f'Toplam {len(all_models)} model yüklendi')
```

```python
def full_metrics(model, X, y, label=''):
    y_pred = model.predict(X)
    try:
        y_prob = model.predict_proba(X)[:, 1]
        auc = round(roc_auc_score(y, y_prob), 4)
    except:
        auc = 'N/A'
    return {
        'Scenario': label,
        'Accuracy':  round(accuracy_score(y, y_pred), 4),
        'Precision': round(precision_score(y, y_pred, zero_division=0), 4),
        'Recall':    round(recall_score(y, y_pred, zero_division=0), 4),
        'F1':        round(f1_score(y, y_pred, zero_division=0), 4),
        'AUC':       auc,
    }

print('Metrik hesaplama fonksiyonu hazır')
```

## 📊 Akademik Özet Tablo

```python
print_section('FULL EVALUATION MATRIX')

rows = []
adversarial_sets = {
    'Normal':   X_test,
    'Minimal':  X_adv_minimal,
    'Gaussian': X_adv_gaussian,
    'Combined': X_adv_combined,
}

for model_name, model in all_models.items():
    for adv_name, X_adv in adversarial_sets.items():
        m = full_metrics(model, X_adv, y_test, label=adv_name)
        m['Model'] = model_name
        rows.append(m)

full_df = pd.DataFrame(rows)
pivot_acc = full_df.pivot_table(values='Accuracy', index='Model', columns='Scenario')
pivot_f1  = full_df.pivot_table(values='F1', index='Model', columns='Scenario')

print('\nACCURACY (Normal | Minimal | Gaussian | Combined):')
print(pivot_acc[['Normal', 'Minimal', 'Gaussian', 'Combined']].to_string())

print('\nF1 SCORE:')
print(pivot_f1[['Normal', 'Minimal', 'Gaussian', 'Combined']].to_string())
```

```python
# Heatmap - final akademik görselleştirme
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

col_order = ['Normal', 'Minimal', 'Gaussian', 'Combined']

for ax, (metric, pivot) in zip(axes, [('Accuracy', pivot_acc), ('F1', pivot_f1)]):
    available_cols = [c for c in col_order if c in pivot.columns]
    sns.heatmap(
        pivot[available_cols], annot=True, fmt='.3f',
        cmap='RdYlGn', vmin=0.5, vmax=1.0, ax=ax,
        linewidths=0.5, linecolor='white',
        annot_kws={'size': 10, 'weight': 'bold'}
    )
    ax.set_title(f'{metric} Karşılaştırması\n(Satır=Model, Sütun=Saldırı Türü)', fontsize=12)
    ax.set_xlabel('Saldırı Senaryosu')
    ax.set_ylabel('Model')

plt.suptitle('Final Evaluation Matrix: Tüm Modeller × Tüm Saldırılar',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(FIGURES_DIR / 'final_evaluation_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Robustness sıralaması
print_section('ROBUSTNESS SIRALAMASI')

robustness_rows = []
for model_name, model in all_models.items():
    acc_normal   = accuracy_score(y_test, model.predict(X_test))
    acc_combined = accuracy_score(y_test, model.predict(X_adv_combined))
    drop         = acc_normal - acc_combined
    
    robustness_rows.append({
        'Model':           model_name,
        'Normal Acc':      round(acc_normal, 4),
        'Adversarial Acc': round(acc_combined, 4),
        'Accuracy Drop':   round(drop, 4),
        'Robustness':      round(1 - drop, 4),
        'Type':            'Defended' if '-D' in model_name or model_name == 'ENS' else 'Baseline'
    })

rob_df = pd.DataFrame(robustness_rows).sort_values('Robustness', ascending=False)
print(rob_df.to_string(index=False))
rob_df.to_csv('../data/processed/robustness_ranking.csv', index=False)
```

```python
# Sonuç grafiği
fig, ax = plt.subplots(figsize=(12, 6))

colors = {'Baseline': '#FF9800', 'Defended': '#2196F3'}
bar_colors = [colors.get(t, '#9E9E9E') for t in rob_df['Type']]

x = np.arange(len(rob_df))
width = 0.35

ax.bar(x - width/2, rob_df['Normal Acc'], width, label='Normal', 
       color='#4CAF50', alpha=0.7)
ax.bar(x + width/2, rob_df['Adversarial Acc'], width, label='Adversarial',
       color=bar_colors, alpha=0.85)

ax.set_xticks(x)
ax.set_xticklabels(rob_df['Model'], rotation=15, fontsize=10)
ax.set_ylim(0, 1.15)
ax.set_ylabel('Accuracy')
ax.set_title('Final Karşılaştırma: Normal vs Adversarial Accuracy\n(Turuncu=Baseline, Mavi=Defended)', 
             fontsize=13)
ax.legend()
ax.grid(axis='y', alpha=0.3)

# Değerleri üstüne yaz
for xi, (_, row) in enumerate(rob_df.iterrows()):
    drop = row['Accuracy Drop']
    ax.text(xi + width/2, row['Adversarial Acc'] - 0.04, f'↓{drop:.3f}',
            ha='center', fontsize=8, color='darkred', fontweight='bold')

plt.tight_layout()
plt.savefig(FIGURES_DIR / 'final_robustness_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```
