# 📓 Notebook 6: SHAP Explainability

Bu notebook'ta:
- Her model için SHAP analizi yapacağız
- Adversarial saldırı öncesi/sonrası SHAP karşılaştırması yapacağız
- Tekil örnek açıklamaları üretiyor olacağız
- 'Neden phishing?' sorusunu cevaplayacağız


```python
import sys
sys.path.append('..')

import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
import seaborn as sns
import shap
import warnings
warnings.filterwarnings('ignore')

from src.utils import load_model, print_section, FIGURES_DIR

# SHAP görselleştirmeleri için arka plan ayarı
shap.initjs()
print(f'✅ SHAP version: {shap.__version__}')
```

```python
# Verileri yükle
X_train = np.load('../data/processed/X_train.npy')
X_test  = np.load('../data/processed/X_test.npy')
y_train = np.load('../data/processed/y_train.npy')
y_test  = np.load('../data/processed/y_test.npy')
X_adv_combined = np.load('../data/processed/X_adv_combined.npy')

with open('../data/processed/feature_names.json') as f:
    feature_names = json.load(f)

# Modelleri yükle
rf_model  = load_model('baseline_random_forest')
xgb_model = load_model('baseline_xgboost')

print(f'Veriler yüklendi | Test: {X_test.shape} | Features: {len(feature_names)}')
```

## 🌲 Random Forest SHAP Analizi

```python
print_section('Random Forest SHAP Analizi')

# TreeExplainer - RF ve XGBoost için hızlı SHAP
print('SHAP explainer oluşturuluyor...')
rf_explainer = shap.TreeExplainer(rf_model)

# Test setinin bir kısmı üzerinde SHAP hesapla (hız için)
sample_size = min(500, len(X_test))
idx_sample  = np.random.choice(len(X_test), sample_size, replace=False)
X_sample    = X_test[idx_sample]
y_sample    = y_test[idx_sample]

print(f'SHAP değerleri hesaplanıyor ({sample_size} örnek)...')
rf_shap_values = rf_explainer.shap_values(X_sample)

# Sınıf 1 (phishing) için SHAP değerleri
if isinstance(rf_shap_values, list):
    shap_phishing = rf_shap_values[1]  # Phishing sınıfı
else:
    shap_phishing = rf_shap_values

print(f'✅ SHAP hesaplandı: {shap_phishing.shape}')
```

```python
# SHAP Summary Plot - Bar
fig, ax = plt.subplots(figsize=(10, 8))
shap.summary_plot(
    shap_phishing, X_sample,
    feature_names=feature_names,
    plot_type='bar',
    max_display=20,
    show=False
)
plt.title('SHAP Feature Importance - Random Forest (Phishing Sınıfı)', fontsize=13)
plt.tight_layout()
plt.savefig(FIGURES_DIR / 'shap_rf_bar.png', dpi=150, bbox_inches='tight')
plt.show()
print('✅ SHAP bar plot kaydedildi')
```

```python
# SHAP Beeswarm Plot (daha detaylı)
fig, ax = plt.subplots(figsize=(10, 10))
shap.summary_plot(
    shap_phishing, X_sample,
    feature_names=feature_names,
    max_display=20,
    show=False
)
plt.title('SHAP Beeswarm - Random Forest', fontsize=13)
plt.tight_layout()
plt.savefig(FIGURES_DIR / 'shap_rf_beeswarm.png', dpi=150, bbox_inches='tight')
plt.show()
```

## ⚡ XGBoost SHAP Analizi

```python
print_section('XGBoost SHAP Analizi')

xgb_explainer = shap.TreeExplainer(xgb_model)
xgb_shap_values = xgb_explainer.shap_values(X_sample)

fig, ax = plt.subplots(figsize=(10, 8))
shap.summary_plot(
    xgb_shap_values, X_sample,
    feature_names=feature_names,
    plot_type='bar',
    max_display=20,
    show=False
)
plt.title('SHAP Feature Importance - XGBoost', fontsize=13)
plt.tight_layout()
plt.savefig(FIGURES_DIR / 'shap_xgb_bar.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 🔴 Adversarial SHAP: Saldırı Öncesi vs Sonrası

```python
print_section('Adversarial SHAP Karşılaştırması')
print('Saldırı öncesi ve sonrası feature importance nasıl değişiyor?')

# Sadece phishing örnekleri
phish_idx = np.where(y_sample == 1)[0][:100]
X_phish_normal = X_sample[phish_idx]
X_phish_adv    = X_adv_combined[idx_sample][phish_idx]

# SHAP değerleri
shap_normal = rf_explainer.shap_values(X_phish_normal)
shap_adv    = rf_explainer.shap_values(X_phish_adv)

if isinstance(shap_normal, list):
    shap_normal = shap_normal[1]
    shap_adv    = shap_adv[1]

# Mean absolute SHAP karşılaştırması
mean_normal = np.abs(shap_normal).mean(axis=0)
mean_adv    = np.abs(shap_adv).mean(axis=0)

# Top 15 feature
top_idx = np.argsort(mean_normal)[::-1][:15]
top_features = [feature_names[i] for i in top_idx]
top_normal   = mean_normal[top_idx]
top_adv      = mean_adv[top_idx]

# Görselleştir
fig, ax = plt.subplots(figsize=(12, 7))
x = np.arange(len(top_features))
width = 0.35

bars1 = ax.bar(x - width/2, top_normal, width, label='Normal', color='#2196F3', alpha=0.8)
bars2 = ax.bar(x + width/2, top_adv,    width, label='Adversarial', color='#F44336', alpha=0.8)

ax.set_xlabel('Feature')
ax.set_ylabel('Mean |SHAP|')
ax.set_title('SHAP Karşılaştırması: Adversarial Saldırı Öncesi vs Sonrası\n(Random Forest, Phishing Örnekleri)', fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels(top_features, rotation=45, ha='right', fontsize=9)
ax.legend()
plt.tight_layout()
plt.savefig(FIGURES_DIR / 'shap_adversarial_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

print('✅ Adversarial SHAP karşılaştırması kaydedildi')
```

## 🔍 Tekil Örnek Açıklaması: "Neden Phishing?"

```python
print_section('Tekil Örnek Açıklaması')

# Phishing olarak tahmin edilen bir örnek seç
predictions = rf_model.predict(X_test)
correct_phish_idx = np.where((predictions == 1) & (y_test == 1))[0]

if len(correct_phish_idx) > 0:
    example_idx = correct_phish_idx[0]
    X_example   = X_test[example_idx:example_idx+1]
    
    shap_example = rf_explainer.shap_values(X_example)
    if isinstance(shap_example, list):
        shap_example = shap_example[1]
    
    # Force plot
    print('Tekil örnek SHAP force plot oluşturuluyor...')
    
    # Manuel bar plot (force plot yerine - notebook uyumluluğu için)
    shap_vals = shap_example[0]
    sorted_idx = np.argsort(np.abs(shap_vals))[::-1][:15]
    
    fig, ax = plt.subplots(figsize=(10, 6))
    colors = ['#F44336' if v > 0 else '#4CAF50' for v in shap_vals[sorted_idx]]
    ax.barh(
        [feature_names[i] for i in sorted_idx][::-1],
        shap_vals[sorted_idx][::-1],
        color=colors[::-1], alpha=0.85
    )
    ax.axvline(x=0, color='black', linewidth=0.8)
    ax.set_title('Bu URL Neden Phishing?\nSHAP Feature Katkıları (Kırmızı=Phishing, Yeşil=Meşru)', fontsize=11)
    ax.set_xlabel('SHAP Değeri (Phishing Olasılığına Katkı)')
    plt.tight_layout()
    plt.savefig(FIGURES_DIR / 'shap_single_example.png', dpi=150, bbox_inches='tight')
    plt.show()
    
    # Yazılı açıklama
    print('\n📝 Bu URL Phishing Çünkü:')
    print('-'*50)
    top_reasons = sorted_idx[:5]
    for rank, feat_i in enumerate(top_reasons, 1):
        feat_name  = feature_names[feat_i]
        feat_value = X_example[0, feat_i]
        shap_val   = shap_vals[feat_i]
        direction  = '⬆️ Phishing riski artırıyor' if shap_val > 0 else '⬇️ Phishing riskini azaltıyor'
        print(f'  {rank}. {feat_name}: {feat_value:.3f} → {direction} (SHAP: {shap_val:+.3f})')
```
